```python
import random
import time
from datetime import datetime
import pandas as pd
```

#User Profile

```python
def get_user_profile():
    print("="*60)
    print("🧠 Real Cognitive Screening Test")
    print("="*60)

    name = input("Enter your full name: ").strip()
    age = int(input("Enter your age: ").strip())
    country = input("Which country are you in? ").strip()
    city = input("Which city are you in? ").strip()
    education = input("Education level: ").strip()

    test_date = datetime.now()

    profile = {
        "name": name,
        "age": age,
        "country": country,
        "city": city,
        "education": education,
        "year": test_date.year,
        "test_date": test_date.strftime("%d-%m-%Y %H:%M")
    }

    return profile
```

#Helper Functions

```python
def clean(text):
    return str(text).strip().lower()

def check_text(user, real):
    return clean(user) in clean(real)

def check_age(user_age, real_age):
    try:
        return abs(int(user_age) - int(real_age)) <= 1
    except:
        return False
```

#Orientation Test

```python
def orientation_test(profile):
    print("\n🧭 Orientation Test")
    score = 0
    total = 5
    times = []

    questions = [
        ("What is your full name? ", profile["name"], "text"),
        ("What is your age? ", profile["age"], "age"),
        ("Which country are you in? ", profile["country"], "text"),
        ("Which city are you in? ", profile["city"], "text"),
        ("What year is it now? ", profile["year"], "age")
    ]

    for q, correct, qtype in questions:
        start = time.time()
        ans = input(q)
        end = time.time()
        times.append(end - start)

        if qtype == "text":
            if check_text(ans, correct):
                score += 1
        else:
            if check_age(ans, correct):
                score += 1

    return score, total, times
```

#Immediate Memory Test


```python
def immediate_memory_test():
    print("\n🧠 Memory Test")

    words = random.sample(
        ["apple", "river", "chair", "banana", "table", "green"],
        3
    )

    print("Remember these words:")
    print(", ".join(words))
    print("You will be asked later...\n")

    time.sleep(3)

    return words
```

#Attention Test

```python
def attention_test():
    print("\n🔢 Attention Test")
    score = 0
    total = 3
    times = []

    questions = [
        ("What is 7 + 5? ", "12"),
        ("What is 15 - 6? ", "9"),
        ("What is 9 + 8? ", "17")
    ]

    for q, correct in questions:
        start = time.time()
        ans = input(q)
        end = time.time()
        times.append(end - start)

        if clean(ans) == correct:
            score += 1

    return score, total, times
```

#Delayed Recall

```python
def delayed_recall_test(words):
    print("\n🧠 Delayed Recall")
    print("Please type the three words you remember (separated by comma)")

    start = time.time()
    user_input = input("Your recall: ")
    end = time.time()

    response_time = end - start

    # process user input
    user_words = [w.strip().lower() for w in user_input.split(",")]
    correct_words = [w.lower() for w in words]

    # count correct matches (no duplicate counting)
    correct_count = 0
    matched = set()

    for uw in user_words:
        if uw in correct_words and uw not in matched:
            correct_count += 1
            matched.add(uw)

    total = len(words)
    times = [response_time]

    print(f"✅ You recalled {correct_count} out of {total} words.\n")

    return correct_count, total, times
```

#Risk Calculation

```python
def calculate_risk(score_percent, avg_time, age):

    # base risk
    if score_percent >= 85:
        risk = "Normal ✅"
    elif score_percent >= 70:
        risk = "Mild Concern ⚠️"
    elif score_percent >= 50:
        risk = "Moderate Risk ⚠️"
    else:
        risk = "High Risk 🚨"

    # age adjustment
    if age < 40 and score_percent < 50:
        risk += " (Low confidence — young age)"

    # slow response warning
    if avg_time > 10:
        risk += " | Slow response detected"

    return risk
```

#Report Generator

```python
def generate_report(profile, score_percent, avg_time, risk):

    report = f"""
====================================
🧠 Cognitive Screening Report
====================================

Name: {profile['name']}
Age: {profile['age']}
Education: {profile['education']}
Test Date: {profile['test_date']}

------------------------------------
Score: {score_percent:.2f}%
Average Response Time: {avg_time:.2f} sec
Risk Level: {risk}

⚠️ This is a screening tool, not a medical diagnosis.
====================================
"""

    filename = f"report_{profile['name'].replace(' ','_')}.txt"

    with open(filename, "w", encoding="utf-8") as f:
        f.write(report)

    print(report)
    print("✅ Report saved:", filename)
```

#Main Controller

```python

def main():
    profile = get_user_profile()

    # tests
    o_score, o_total, o_times = orientation_test(profile)
    words = immediate_memory_test()
    a_score, a_total, a_times = attention_test()
    d_score, d_total, d_times = delayed_recall_test(words)

    # combine
    total_score = o_score + a_score + d_score
    total_questions = o_total + a_total + d_total

    all_times = o_times + a_times + d_times
    avg_time = sum(all_times) / len(all_times)

    score_percent = (total_score / total_questions) * 100

    risk = calculate_risk(score_percent, avg_time, profile["age"])

    generate_report(profile, score_percent, avg_time, risk)

# RUN
main()
```
